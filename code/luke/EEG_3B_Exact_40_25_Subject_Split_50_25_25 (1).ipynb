```python
from pathlib import Path
import pandas as pd
import sys
import importlib
import shutil
import os
import time

# ============================================================
# PATHS
# ============================================================

CODE_DIR = Path(
    "/content/drive/MyDrive/EEG Shared/Code/Code Luke"
)

SOURCE_DIR = Path(
    "/content/drive/MyDrive/EEG Shared/Processed_data/"
    "Kosachenko_All_Subjects"
)

OUTPUT_DIR = SOURCE_DIR / "3B_Exact_40_25_Split"

MANIFEST_PATH = OUTPUT_DIR / "subject_assignments.csv"

DRIVE_DIR = OUTPUT_DIR / "subject_holdout"

DRIVE_TRAIN = DRIVE_DIR / "train.csv"
DRIVE_VALIDATION = DRIVE_DIR / "validation.csv"
DRIVE_HOLDOUT = DRIVE_DIR / "holdout.csv"

LOCAL_DIR = Path("/content/3b_holdout_full_rebuild")
LOCAL_DIR.mkdir(parents=True, exist_ok=True)

LOCAL_TRAIN = LOCAL_DIR / "train.csv"
LOCAL_VALIDATION = LOCAL_DIR / "validation.csv"
LOCAL_HOLDOUT = LOCAL_DIR / "holdout.csv"


# ============================================================
# IMPORT ORIGINAL 3B HELPERS
# ============================================================

if str(CODE_DIR) not in sys.path:
    sys.path.insert(0, str(CODE_DIR))

module_name = "load_data_3b_exact_40_25_split_50_25_25"

m = importlib.import_module(module_name)
m = importlib.reload(m)

_read_csv_with_retries = m._read_csv_with_retries
_subject_from_filename = m._subject_from_filename
_normalize_source_frame = m._normalize_source_frame
_split_epoch_ids = m._split_epoch_ids
_append_frame = m._append_frame
find_processed_csvs = m.find_processed_csvs

print("3B functions loaded.")


# ============================================================
# SETTINGS
# ============================================================

RANDOM_STATE = 1

FRACTIONS = {
    "train": 0.50,
    "validation": 0.25,
    "holdout": 0.25,
}


def norm_subject(x):
    s = str(x).strip()

    if s.endswith(".0"):
        s = s[:-2]

    if s.isdigit():
        return str(int(s))

    return s


# ============================================================
# DELETE OLD LOCAL TEMP FILES
# ============================================================

for p in [
    LOCAL_TRAIN,
    LOCAL_VALIDATION,
    LOCAL_HOLDOUT,
]:
    if p.exists():
        p.unlink()


# ============================================================
# GET EXACT ORIGINAL 25 HOLDOUT SUBJECTS
# ============================================================

manifest = pd.read_csv(MANIFEST_PATH)

holdout_subjects = [
    norm_subject(x)
    for x in manifest.loc[
        manifest["subject_split"] == "subject_holdout",
        "subject_id"
    ].tolist()
]

print("Holdout subjects:", len(holdout_subjects))
print([x.zfill(3) for x in holdout_subjects])

if len(holdout_subjects) != 25:
    raise RuntimeError(
        f"Expected 25 holdout subjects, found {len(holdout_subjects)}"
    )


# ============================================================
# FIND ORIGINAL 130 FILES
# ============================================================

source_files = find_processed_csvs(SOURCE_DIR)

print("Original processed files:", len(source_files))

subject_to_files = {}
original_subject_ids = {}

for file_path in source_files:

    parsed = _subject_from_filename(file_path)
    normalized = norm_subject(parsed)

    subject_to_files.setdefault(
        normalized, []
    ).append(file_path)

    original_subject_ids[normalized] = str(parsed)


# ============================================================
# MASTER COLUMN ORDER
# ============================================================

first_frame = _read_csv_with_retries(source_files[0])

master_source_columns = list(first_frame.columns)
master_columns = list(master_source_columns)

for col in ["subject_split", "sample_split"]:
    if col not in master_columns:
        master_columns.append(col)

del first_frame


local_outputs = {
    "train": LOCAL_TRAIN,
    "validation": LOCAL_VALIDATION,
    "holdout": LOCAL_HOLDOUT,
}

epoch_counts = {
    "train": 0,
    "validation": 0,
    "holdout": 0,
}

row_counts = {
    "train": 0,
    "validation": 0,
    "holdout": 0,
}

subjects_written = {
    "train": set(),
    "validation": set(),
    "holdout": set(),
}


# ============================================================
# REBUILD ALL 25 HOLDOUT SUBJECTS FROM ORIGINAL DATA
# ============================================================

for number, subject_id in enumerate(
    holdout_subjects,
    start=1
):

    print(
        f"\nRebuilding holdout "
        f"{number}/25: sub-{subject_id.zfill(3)}"
    )

    if subject_id not in subject_to_files:
        raise RuntimeError(
            f"No source files found for subject {subject_id}"
        )

    frames = []

    for file_path in subject_to_files[subject_id]:

        frame = _read_csv_with_retries(file_path)

        frame = _normalize_source_frame(
            frame,
            file_path,
            master_source_columns,
        )

        frames.append(frame)

    subject_data = pd.concat(
        frames,
        ignore_index=True
    )

    del frames


    # Complete 24-row epochs only
    counts = subject_data.groupby("epoch_uid").size()

    valid_epoch_ids = (
        counts[counts == 24]
        .index
        .tolist()
    )

    subject_data = subject_data[
        subject_data["epoch_uid"].isin(valid_epoch_ids)
    ].copy()


    # EXACT SAME ORIGINAL PER-SUBJECT SEED
    seed_subject = original_subject_ids[subject_id]

    split_seed = (
        RANDOM_STATE
        + sum(ord(c) for c in seed_subject)
    )

    split_ids = _split_epoch_ids(
        valid_epoch_ids,
        FRACTIONS,
        split_seed,
    )


    # Verify assignment
    assigned = [
        epoch
        for ids in split_ids.values()
        for epoch in ids
    ]

    if len(assigned) != len(set(assigned)):
        raise RuntimeError(
            f"Epoch overlap for subject {subject_id}"
        )

    if set(assigned) != set(valid_epoch_ids):
        raise RuntimeError(
            f"Missing epoch assignment for subject {subject_id}"
        )


    # Write all three locally
    for split_name, ids in split_ids.items():

        split_frame = subject_data[
            subject_data["epoch_uid"].isin(ids)
        ].copy()

        split_frame["subject_split"] = "subject_holdout"
        split_frame["sample_split"] = split_name

        _append_frame(
            split_frame,
            local_outputs[split_name],
            master_columns,
        )

        subjects_written[split_name].add(subject_id)
        epoch_counts[split_name] += len(ids)
        row_counts[split_name] += len(split_frame)

        print(
            f"  {split_name}: "
            f"{len(ids)} epochs / "
            f"{len(split_frame)} rows"
        )

    del subject_data


# ============================================================
# VERIFY WITHOUT CHUNK-BOUNDARY FALSE ERRORS
# ============================================================

print("\n" + "=" * 70)
print("LOCAL HOLDOUT REBUILD COMPLETE")
print("=" * 70)

for split_name in [
    "train",
    "validation",
    "holdout",
]:

    subjects = subjects_written[split_name]
    epochs = epoch_counts[split_name]
    rows = row_counts[split_name]

    print(
        f"{split_name}: "
        f"{len(subjects)} subjects | "
        f"{epochs} epochs | "
        f"{rows} rows"
    )

    if len(subjects) != 25:
        raise RuntimeError(
            f"{split_name}: expected 25 subjects, "
            f"got {len(subjects)}"
        )

    if rows != epochs * 24:
        raise RuntimeError(
            f"{split_name}: rows != epochs × 24"
        )

print("\nPASS — all 25 subjects rebuilt correctly.")


# ============================================================
# COPY COMPLETE FILES TO DRIVE
# ============================================================

DRIVE_DIR.mkdir(
    parents=True,
    exist_ok=True
)

pairs = [
    (LOCAL_TRAIN, DRIVE_TRAIN),
    (LOCAL_VALIDATION, DRIVE_VALIDATION),
    (LOCAL_HOLDOUT, DRIVE_HOLDOUT),
]

print("\nCopying completed files to Drive...")

for local_path, drive_path in pairs:

    shutil.copy2(
        local_path,
        drive_path
    )

    os.sync()
    time.sleep(3)

    if not drive_path.exists():
        raise RuntimeError(
            f"Missing after copy: {drive_path}"
        )

    if (
        local_path.stat().st_size
        != drive_path.stat().st_size
    ):
        raise RuntimeError(
            f"Size mismatch: {drive_path}"
        )

    print(
        drive_path.name,
        "PASS |",
        round(
            drive_path.stat().st_size / (1024 ** 3),
            3
        ),
        "GB"
    )


print("\n" + "=" * 70)
print("3B HOLDOUT REBUILD COMPLETE")
print("=" * 70)

print(
    "All 25 holdout subjects now have "
    "train / validation / holdout data."
)
```

```python
# ============================================================
# FULL 3B DEVELOPMENT REPAIR
#
# Rebuilds:
#   subject_train/train.csv
#   subject_train/validation.csv
#   subject_train/holdout.csv
#
# Leaves subject_holdout completely untouched.
# Builds locally first, verifies, THEN copies to Google Drive.
# ============================================================

from pathlib import Path
import pandas as pd
import numpy as np
import sys
import importlib
import shutil
import os
import time


# ============================================================
# PATHS
# ============================================================

CODE_DIR = Path(
    "/content/drive/MyDrive/EEG Shared/Code/Code Luke"
)

SOURCE_DIR = Path(
    "/content/drive/MyDrive/EEG Shared/Processed_data/"
    "Kosachenko_All_Subjects"
)

OUTPUT_DIR = (
    SOURCE_DIR /
    "3B_Exact_40_25_Split"
)

MANIFEST_PATH = (
    OUTPUT_DIR /
    "subject_assignments.csv"
)

DRIVE_TRAIN = (
    OUTPUT_DIR /
    "subject_train" /
    "train.csv"
)

DRIVE_VALIDATION = (
    OUTPUT_DIR /
    "subject_train" /
    "validation.csv"
)

DRIVE_HOLDOUT = (
    OUTPUT_DIR /
    "subject_train" /
    "holdout.csv"
)


# Build all 3 locally first
LOCAL_DIR = Path(
    "/content/3b_development_repair"
)

LOCAL_DIR.mkdir(
    parents=True,
    exist_ok=True
)

LOCAL_TRAIN = (
    LOCAL_DIR /
    "train.csv"
)

LOCAL_VALIDATION = (
    LOCAL_DIR /
    "validation.csv"
)

LOCAL_HOLDOUT = (
    LOCAL_DIR /
    "holdout.csv"
)


# ============================================================
# LOAD ORIGINAL 3B FUNCTIONS
# ============================================================

if str(CODE_DIR) not in sys.path:
    sys.path.insert(
        0,
        str(CODE_DIR)
    )

module_name = (
    "load_data_3b_exact_40_25_split_50_25_25"
)

m = importlib.import_module(
    module_name
)

m = importlib.reload(m)

_read_csv_with_retries = (
    m._read_csv_with_retries
)

_subject_from_filename = (
    m._subject_from_filename
)

_normalize_source_frame = (
    m._normalize_source_frame
)

_split_epoch_ids = (
    m._split_epoch_ids
)

_append_frame = (
    m._append_frame
)

find_processed_csvs = (
    m.find_processed_csvs
)

print("3B functions loaded.")


# ============================================================
# SETTINGS
# ============================================================

RANDOM_STATE = 1

FRACTIONS = {
    "train": 0.50,
    "validation": 0.25,
    "holdout": 0.25,
}


# ============================================================
# SUBJECT NORMALIZATION
# ============================================================

def normalize_subject_id(value):

    s = str(value).strip()

    if s.endswith(".0"):
        s = s[:-2]

    if s.isdigit():
        return str(int(s))

    return s


# ============================================================
# REMOVE OLD LOCAL REPAIR FILES
# ============================================================

for path in [
    LOCAL_TRAIN,
    LOCAL_VALIDATION,
    LOCAL_HOLDOUT,
]:

    if path.exists():
        path.unlink()


# ============================================================
# LOAD EXACT 40 DEVELOPMENT SUBJECT ASSIGNMENTS
# ============================================================

if not MANIFEST_PATH.exists():

    raise FileNotFoundError(
        f"Missing manifest:\n"
        f"{MANIFEST_PATH}"
    )


manifest = pd.read_csv(
    MANIFEST_PATH
)


development_subjects = [

    normalize_subject_id(x)

    for x in manifest.loc[
        manifest[
            "subject_split"
        ] == "subject_train",
        "subject_id"
    ].tolist()
]


print(
    "Development subjects:",
    len(development_subjects)
)


if len(development_subjects) != 40:

    raise RuntimeError(
        "Expected exactly 40 "
        "development subjects."
    )


# ============================================================
# FIND ORIGINAL 130 PROCESSED FILES
# ============================================================

source_files = (
    find_processed_csvs(
        SOURCE_DIR
    )
)

print(
    "Original processed files:",
    len(source_files)
)


if len(source_files) != 130:

    print(
        "WARNING: expected 130 "
        "processed files, found",
        len(source_files)
    )


# ============================================================
# MAP ORIGINAL FILES TO SUBJECTS
#
# Preserve original zero-padded ID for deterministic seed:
# 51 -> original ID "051"
# ============================================================

subject_to_files = {}

original_subject_id = {}


for file_path in source_files:

    parsed = _subject_from_filename(
        file_path
    )

    normalized = normalize_subject_id(
        parsed
    )

    subject_to_files.setdefault(
        normalized,
        []
    ).append(file_path)

    original_subject_id[
        normalized
    ] = str(parsed)


print(
    "Subjects found in originals:",
    len(subject_to_files)
)


# ============================================================
# MASTER COLUMN SCHEMA
# ============================================================

first_frame = (
    _read_csv_with_retries(
        source_files[0]
    )
)

master_source_columns = list(
    first_frame.columns
)

master_columns = list(
    master_source_columns
)

for column in [
    "subject_split",
    "sample_split",
]:

    if column not in master_columns:
        master_columns.append(
            column
        )

del first_frame


# ============================================================
# OUTPUT PATH MAP
# ============================================================

local_outputs = {

    "train":
        LOCAL_TRAIN,

    "validation":
        LOCAL_VALIDATION,

    "holdout":
        LOCAL_HOLDOUT,
}


# ============================================================
# COUNTS FOR VERIFICATION
# ============================================================

epoch_counts = {

    "train": 0,
    "validation": 0,
    "holdout": 0,
}

row_counts = {

    "train": 0,
    "validation": 0,
    "holdout": 0,
}

subjects_written = {

    "train": set(),
    "validation": set(),
    "holdout": set(),
}


# ============================================================
# REBUILD ALL 40 DEVELOPMENT SUBJECTS
# ============================================================

for number, subject_id in enumerate(
    development_subjects,
    start=1
):

    print(
        f"\nRepairing "
        f"{number}/40: "
        f"{subject_id.zfill(3)}"
    )

    if subject_id not in subject_to_files:

        raise RuntimeError(
            f"No source files for "
            f"subject {subject_id}"
        )


    # --------------------------------------------------------
    # LOAD THIS SUBJECT ONLY
    # --------------------------------------------------------

    frames = []

    for file_path in (
        subject_to_files[
            subject_id
        ]
    ):

        frame = (
            _read_csv_with_retries(
                file_path
            )
        )

        frame = (
            _normalize_source_frame(
                frame,
                file_path,
                master_source_columns,
            )
        )

        frames.append(
            frame
        )


    subject_data = pd.concat(
        frames,
        ignore_index=True
    )

    del frames


    # --------------------------------------------------------
    # COMPLETE EPOCHS ONLY
    # --------------------------------------------------------

    counts = (
        subject_data
        .groupby(
            "epoch_uid"
        )
        .size()
    )

    valid_epoch_ids = (
        counts[
            counts == 24
        ]
        .index
        .tolist()
    )


    incomplete = int(
        (
            counts != 24
        ).sum()
    )

    if incomplete:

        print(
            "  dropping",
            incomplete,
            "incomplete epochs"
        )


    subject_data = (
        subject_data[
            subject_data[
                "epoch_uid"
            ].isin(
                valid_epoch_ids
            )
        ]
        .copy()
    )


    # --------------------------------------------------------
    # EXACT SAME DETERMINISTIC SPLIT
    # AS ORIGINAL 3B
    # --------------------------------------------------------

    seed_subject = (
        original_subject_id[
            subject_id
        ]
    )

    split_seed = (
        RANDOM_STATE
        +
        sum(
            ord(c)
            for c in seed_subject
        )
    )


    split_ids = (
        _split_epoch_ids(
            valid_epoch_ids,
            FRACTIONS,
            split_seed,
        )
    )


    # --------------------------------------------------------
    # VERIFY NO OVERLAP
    # --------------------------------------------------------

    all_assigned = [

        epoch

        for ids
        in split_ids.values()

        for epoch
        in ids
    ]


    if (
        len(all_assigned)
        !=
        len(
            set(all_assigned)
        )
    ):

        raise RuntimeError(
            f"Epoch overlap for "
            f"subject {subject_id}"
        )


    if (
        set(all_assigned)
        !=
        set(valid_epoch_ids)
    ):

        raise RuntimeError(
            f"Not every epoch assigned "
            f"for subject {subject_id}"
        )


    # --------------------------------------------------------
    # WRITE TRAIN / VALIDATION / HOLDOUT LOCALLY
    # --------------------------------------------------------

    for split_name in [
        "train",
        "validation",
        "holdout",
    ]:

        ids = (
            split_ids[
                split_name
            ]
        )


        split_frame = (
            subject_data[
                subject_data[
                    "epoch_uid"
                ].isin(ids)
            ]
            .copy()
        )


        split_frame[
            "subject_split"
        ] = "subject_train"

        split_frame[
            "sample_split"
        ] = split_name


        # Every epoch must still be 24 rows
        split_sizes = (
            split_frame
            .groupby(
                "epoch_uid"
            )
            .size()
        )

        bad_sizes = (
            split_sizes[
                split_sizes != 24
            ]
        )

        if len(bad_sizes):

            raise RuntimeError(
                f"{subject_id} / "
                f"{split_name}: "
                f"bad epoch sizes"
            )


        _append_frame(
            split_frame,
            local_outputs[
                split_name
            ],
            master_columns,
        )


        epoch_counts[
            split_name
        ] += len(ids)

        row_counts[
            split_name
        ] += len(
            split_frame
        )

        subjects_written[
            split_name
        ].add(
            subject_id
        )


        print(
            f"  {split_name}: "
            f"{len(ids)} epochs / "
            f"{len(split_frame)} rows"
        )


    del subject_data


# ============================================================
# LOCAL VERIFICATION
# ============================================================

print(
    "\n"
    + "=" * 70
)

print(
    "LOCAL REBUILD COMPLETE"
)

print(
    "=" * 70
)


for split_name in [
    "train",
    "validation",
    "holdout",
]:

    print(
        f"\n{split_name.upper()}"
    )

    print(
        "Subjects:",
        len(
            subjects_written[
                split_name
            ]
        )
    )

    print(
        "Epochs:",
        epoch_counts[
            split_name
        ]
    )

    print(
        "Rows:",
        row_counts[
            split_name
        ]
    )

    print(
        "Expected rows:",
        epoch_counts[
            split_name
        ] * 24
    )


    if (
        len(
            subjects_written[
                split_name
            ]
        )
        != 40
    ):

        raise RuntimeError(
            f"{split_name} does not "
            f"contain all 40 subjects."
        )


    if (
        row_counts[
            split_name
        ]
        !=
        epoch_counts[
            split_name
        ] * 24
    ):

        raise RuntimeError(
            f"{split_name} row count "
            f"is not epochs × 24."
        )


    if not (
        local_outputs[
            split_name
        ].exists()
    ):

        raise RuntimeError(
            f"Local {split_name}.csv "
            f"was not created."
        )


print(
    "\nPASS — all 3 local "
    "development files verified."
)


# ============================================================
# COPY COMPLETED FILES TO GOOGLE DRIVE
# ============================================================

print(
    "\nCopying completed files "
    "to Google Drive..."
)


DRIVE_TRAIN.parent.mkdir(
    parents=True,
    exist_ok=True
)


copy_pairs = [

    (
        LOCAL_TRAIN,
        DRIVE_TRAIN
    ),

    (
        LOCAL_VALIDATION,
        DRIVE_VALIDATION
    ),

    (
        LOCAL_HOLDOUT,
        DRIVE_HOLDOUT
    ),
]


for local_path, drive_path in (
    copy_pairs
):

    print(
        "\nCopying:",
        local_path.name
    )

    print(
        "Local size:",
        local_path.stat().st_size
    )


    shutil.copy2(
        local_path,
        drive_path
    )


    os.sync()
    time.sleep(3)


    if not drive_path.exists():

        raise RuntimeError(
            f"Drive copy missing: "
            f"{drive_path}"
        )


    local_size = (
        local_path
        .stat()
        .st_size
    )

    drive_size = (
        drive_path
        .stat()
        .st_size
    )


    print(
        "Drive size:",
        drive_size
    )


    if (
        local_size
        !=
        drive_size
    ):

        raise RuntimeError(
            f"Size mismatch for "
            f"{drive_path.name}"
        )


    print(
        "  PASS"
    )


# ============================================================
# FINAL
# ============================================================

print(
    "\n"
    + "=" * 70
)

print(
    "3B DEVELOPMENT REPAIR COMPLETE"
)

print(
    "=" * 70
)

print(
    "All 40 development subjects "
    "are now present in:"
)

print(
    "  train.csv"
)

print(
    "  validation.csv"
)

print(
    "  holdout.csv"
)

print(
    "\nAll epochs remain "
    "exactly 24 rows."
)

print(
    "\nYou can now rerun 3C."
)
```

```python
from pathlib import Path
import pandas as pd

# ============================================================
# 3B REPAIR — recreate ONLY subject_train/train.csv
# ============================================================

SOURCE_DIR = Path(
    "/content/drive/MyDrive/EEG Shared/Processed_data/Kosachenko_All_Subjects"
)

OUTPUT_DIR = SOURCE_DIR / "3B_Exact_40_25_Split"
MANIFEST_PATH = OUTPUT_DIR / "subject_assignments.csv"
TARGET_PATH = OUTPUT_DIR / "subject_train" / "train.csv"

RANDOM_STATE = 1
TRAIN_FRACTION = 0.50
VALIDATION_FRACTION = 0.25


# ------------------------------------------------------------
# Normalize subject IDs
# Makes 051, 51, 51.0 all become "51"
# ------------------------------------------------------------

def normalize_subject_id(value):
    value = str(value).strip().replace(".0", "")

    if value.isdigit():
        return str(int(value))

    return value


# ------------------------------------------------------------
# Safety checks
# ------------------------------------------------------------

if TARGET_PATH.exists():
    raise FileExistsError(
        f"{TARGET_PATH} already exists.\n"
        "Repair stopped so nothing gets duplicated."
    )

if not MANIFEST_PATH.exists():
    raise FileNotFoundError(
        f"Missing subject assignment manifest:\n{MANIFEST_PATH}"
    )


# ------------------------------------------------------------
# Load exact original subject assignments
# ------------------------------------------------------------

manifest = pd.read_csv(MANIFEST_PATH)

development_subjects = [
    normalize_subject_id(x)
    for x in manifest.loc[
        manifest["subject_split"] == "subject_train",
        "subject_id"
    ].tolist()
]

print("Development subjects found:", len(development_subjects))

if len(development_subjects) != 40:
    raise RuntimeError(
        f"Expected 40 development subjects, "
        f"found {len(development_subjects)}"
    )


# ------------------------------------------------------------
# Find original processed CSVs
# ------------------------------------------------------------

all_processed = sorted(SOURCE_DIR.rglob("*_processed.csv"))

# Ignore any 3B output directories
source_files = [
    p for p in all_processed
    if not any(part.startswith("3B_") for part in p.parts)
]

print("Source processed files found:", len(source_files))

if len(source_files) == 0:
    raise RuntimeError("No original processed CSV files were found.")


# ------------------------------------------------------------
# Match source files to subjects
# ------------------------------------------------------------

subject_to_files = {}

for file_path in source_files:

    try:
        parsed_subject = _subject_from_filename(file_path)
    except Exception:
        continue

    parsed_subject = normalize_subject_id(parsed_subject)

    subject_to_files.setdefault(
        parsed_subject, []
    ).append(file_path)


print("Subjects found in source files:", len(subject_to_files))


# ------------------------------------------------------------
# Get same schema used by original 3B
# ------------------------------------------------------------

first_source = source_files[0]

first_frame = _read_csv_with_retries(first_source)

master_source_columns = list(first_frame.columns)
master_columns = list(master_source_columns)

for column in ["subject_split", "sample_split"]:
    if column not in master_columns:
        master_columns.append(column)

del first_frame


# ------------------------------------------------------------
# Create target directory
# ------------------------------------------------------------

TARGET_PATH.parent.mkdir(
    parents=True,
    exist_ok=True
)


# ------------------------------------------------------------
# Rebuild TRAIN ONLY for the 40 development subjects
# ------------------------------------------------------------

total_epochs = 0
total_rows = 0

for number, subject_id in enumerate(
    development_subjects,
    start=1
):

    subject_id = normalize_subject_id(subject_id)

    if subject_id not in subject_to_files:
        raise RuntimeError(
            f"No original processed files found "
            f"for subject {subject_id}"
        )

    print(
        f"\nRepairing subject "
        f"{number}/40: {subject_id}"
    )

    frames = []

    for file_path in subject_to_files[subject_id]:

        frame = _read_csv_with_retries(
            file_path
        )

        frame = _normalize_source_frame(
            frame,
            file_path,
            master_source_columns,
        )

        frames.append(frame)

    if not frames:
        raise RuntimeError(
            f"No readable files for subject {subject_id}"
        )

    subject_data = pd.concat(
        frames,
        ignore_index=True
    )

    del frames


    # --------------------------------------------------------
    # Keep only complete 24-row epochs
    # --------------------------------------------------------

    counts = (
        subject_data
        .groupby("epoch_uid")
        .size()
    )

    valid_epoch_ids = (
        counts[counts == 24]
        .index
        .tolist()
    )

    incomplete = int(
        (counts != 24).sum()
    )

    if incomplete:
        print(
            f"  Dropping {incomplete} "
            f"incomplete epochs"
        )

    subject_data = subject_data[
        subject_data["epoch_uid"].isin(
            valid_epoch_ids
        )
    ].copy()


    # --------------------------------------------------------
    # SAME 50 / 25 / 25 split as original 3B
    # --------------------------------------------------------

    fractions = {
        "train": TRAIN_FRACTION,
        "validation": VALIDATION_FRACTION,
        "holdout":
            1.0
            - TRAIN_FRACTION
            - VALIDATION_FRACTION,
    }

    # IMPORTANT:
    # Use the zero-padded subject ID for the seed.
    #
    # Original 3B used IDs such as "051", not "51",
    # when calculating the deterministic seed.
    original_seed_subject = str(subject_id).zfill(3)

    split_ids = _split_epoch_ids(
        valid_epoch_ids,
        fractions,
        RANDOM_STATE
        + sum(
            ord(c)
            for c in original_seed_subject
        ),
    )

    train_ids = split_ids["train"]


    # --------------------------------------------------------
    # Select training epochs
    # --------------------------------------------------------

    train_frame = subject_data[
        subject_data["epoch_uid"].isin(
            train_ids
        )
    ].copy()

    train_frame["subject_split"] = (
        "subject_train"
    )

    train_frame["sample_split"] = "train"


    # --------------------------------------------------------
    # Append ONLY to missing train.csv
    # --------------------------------------------------------

    _append_frame(
        train_frame,
        TARGET_PATH,
        master_columns,
    )

    total_epochs += len(train_ids)
    total_rows += len(train_frame)

    print(
        f"  restored train: "
        f"{len(train_ids)} epochs / "
        f"{len(train_frame)} rows"
    )

    del subject_data
    del train_frame


# ============================================================
# Verification
# ============================================================

print("\n" + "=" * 70)
print("REPAIR COMPLETE")
print("=" * 70)

print("Created:", TARGET_PATH)
print("Subjects expected:", 40)
print("Training epochs:", total_epochs)
print("Rows:", total_rows)


dtype_map = {
    column: "string"
    for column in [
        "subject_id",
        "subject",
        "epoch_uid",
    ]
    if column in master_columns
}

check = pd.read_csv(
    TARGET_PATH,
    low_memory=False,
    dtype=dtype_map if dtype_map else None,
)


# Check columns
if list(check.columns) != master_columns:
    raise RuntimeError(
        "Column schema does not match "
        "the original 3B schema."
    )


# Determine subject column
subject_col = (
    "subject_id"
    if "subject_id" in check.columns
    else "subject"
)


actual_subjects = {
    normalize_subject_id(x)
    for x in (
        check[subject_col]
        .dropna()
        .tolist()
    )
}

expected_subjects = set(
    development_subjects
)


missing_subjects = (
    expected_subjects
    - actual_subjects
)

unexpected_subjects = (
    actual_subjects
    - expected_subjects
)


print("\nVerification:")
print(
    "Subjects in repaired file:",
    len(actual_subjects)
)

print(
    "Expected subjects:",
    len(expected_subjects)
)


if missing_subjects:
    raise RuntimeError(
        f"Missing subjects: "
        f"{sorted(missing_subjects)}"
    )

if unexpected_subjects:
    raise RuntimeError(
        f"Unexpected subjects: "
        f"{sorted(unexpected_subjects)}"
    )


# Check epoch sizes
epoch_sizes = (
    check
    .groupby("epoch_uid")
    .size()
)

bad_epochs = (
    epoch_sizes[
        epoch_sizes != 24
    ]
)

print(
    "Epochs:",
    len(epoch_sizes)
)

print(
    "Bad epoch sizes:",
    len(bad_epochs)
)


if len(bad_epochs) != 0:
    raise RuntimeError(
        f"{len(bad_epochs)} epochs "
        f"do not contain exactly 24 rows."
    )


# Check split labels
if set(check["subject_split"].dropna().unique()) != {
    "subject_train"
}:
    raise RuntimeError(
        "Incorrect subject_split labels."
    )

if set(check["sample_split"].dropna().unique()) != {
    "train"
}:
    raise RuntimeError(
        "Incorrect sample_split labels."
    )


print("\n" + "=" * 70)
print(
    "PASS — subject_train/train.csv "
    "successfully restored."
)
print("=" * 70)
```
